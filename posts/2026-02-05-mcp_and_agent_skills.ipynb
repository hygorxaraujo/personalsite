# Model Context Protocol and Agent Skills

With the increase in adoption of [Large Language Models (LLMs)](https://en.wikipedia.org/wiki/Large_language_model) and the development of AI Agents based on them, there is a need of enhancing those agents with specialized capabilities as well as some general ones. To enable the creation of those capabilities in a structured and reusable way [Anthropic](https://en.wikipedia.org/wiki/Anthropic) developed the Model Context Protocol (MCP) defining a communication standard to connect agents to systems, data and tools.

More recently, Anthropic also created a standard for the development of Agent Skills, which is a form of packaging instructions and resources for agents to perform specific tasks.

Let's understand what are the MCP and the Agent Skills standards, their differences and use cases.

## Model Context Protocol (MCP)


### Architecture

MCP follows a client-server architecture
The main components of the architecture are:
- MCP Host
- MCP Client
- MCP Server

```{mermaid}

    subgraph host ["AI Application (MCP Host)"]
        mcpclient1["MCP Client 1"]
        mcpclient2["MCP Client 2"]
        mcpclient3["MCP Client 3"]
        mcpclient4["MCP Client 4"]
    end

    mcpclient1 -->|"Dedicated connection"| mcpserver1["MCP Server 1"]
    mcpclient2 -->|"Dedicated connection"| mcpserver2["MCP Server 2"]
    mcpclient3 -->|"Dedicated connection"| mcpserver3["MCP Server 3"]
    mcpclient4 -->|"Dedicated connection"| mcpserver3["MCP Server 3"]
```

Has two layers:
- Data layer:
    - Lifecycle management
    - Server features
    - Client features
    - Utility features
- Transport layer:
    - Stdio transport
    - Streamable HTTP transport

#### Server

Responsible for serving context data
can run locally stdio or remotely http which can also be streamable

- Primitives
  - tools
  - resources
  - prompts

- notifications
- tasks

#### Client

unique by connection to mcp server

- Primitives
  - sampling
  - elicitation
  - logging

#### Example

Uses the JSON-RPC protocol 2.0 for the client-server communication.
JSON-RPC is a lightweight remote procedure call protocol.

1. Initialize
2. Request primitives
3. Execute primitives
4. Real-time notifications

```{mermaid}
sequenceDiagram
    participant client as MCP Client
    participant server as MCP Server
    client->>+server: initialize
    server-->>client: initialize_response
    client->>server: notifications/initialized
```

Initialize request:
```json
{
  "jsonrpc": "2.0",
  "id": 1,
  "method": "initialize",
  "params": {
    "protocolVersion": "2025-06-18",
    "capabilities": {
      "elicitation": {}
    },
    "clientInfo": {
      "name": "example-client",
      "version": "1.0.0"
    }
  }
}
```

Initialize response:
```json
{
  "jsonrpc": "2.0",
  "id": 1,
  "result": {
    "protocolVersion": "2025-06-18",
    "capabilities": {
      "tools": {
        "listChanged": true
      },
      "resources": {}
    },
    "serverInfo": {
      "name": "example-server",
      "version": "1.0.0"
    }
  }
}
```

Tools list request:
```json
{
  "jsonrpc": "2.0",
  "id": 2,
  "method": "tools/list" // "resources/list" "prompts/list"
}
```

Tool call request:
```json
{
  "jsonrpc": "2.0",
  "id": 3,
  "method": "tools/call",
  "params": {
    "name": "weather_current",
    "arguments": {
      "location": "San Francisco",
      "units": "imperial"
    }
  }
}
```

## Agent Skills

A skills-compatible agent needs to:

1. Discover skills in configured directories
2. Load metadata (name and description) at startup
3. Match user tasks to relevant skills
4. Activate skills by loading full instructions
5. Execute scripts and access resources as needed

The SKILL.md must have a yaml frontmatter (key-value pair metadata) at the top with name and description, and the skill instructions.

```text
my-skill/
├── SKILL.md          # Required: instructions + metadata
├── scripts/          # Optional: executable code
├── references/       # Optional: documentation
└── assets/           # Optional: templates, resources
```

```markdown
---
name: pdf-processing
description: Extract text and tables from PDF files, fill forms, merge documents.
---

# PDF Processing

## When to use this skill
Use this skill when the user needs to work with PDF files...

## How to extract text
1. Use pdfplumber for text extraction...

## How to fill forms
...
```
